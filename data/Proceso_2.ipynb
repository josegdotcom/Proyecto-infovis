# Procesamiento de datos Prueba 2
## Los datos sobre la informacion de giroscopio de los los telefonos
el obetivo en esta parte sera tomar estos datos y pasarlos a un entorno en donde se pueda utilizar por los modelos.

[Link](https://archive.ics.uci.edu/dataset/240/human%2Bactivity%2Brecognition%2Busing%2Bsmartphones)

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
CARPETA_DATOS = Path(
    "datos/human+activity+recognition+using+smartphones/"
    "UCI HAR Dataset/UCI HAR Dataset"
)

ARCHIVO_SALIDA = Path("datos/har_dataset.npz")
```

### CARGAR DATOS

```python
X_train = pd.read_csv(
    CARPETA_DATOS / "train" / "X_train.txt",
    sep=r"\s+",
    header=None
).values

y_train = pd.read_csv(
    CARPETA_DATOS / "train" / "y_train.txt",
    sep=r"\s+",
    header=None
).values.ravel()


X_test = pd.read_csv(
    CARPETA_DATOS / "test" / "X_test.txt",
    sep=r"\s+",
    header=None
).values

y_test = pd.read_csv(
    CARPETA_DATOS / "test" / "y_test.txt",
    sep=r"\s+",
    header=None
).values.ravel()
```

### CONVERTIR TIPOS

```python
X_train = X_train.astype(np.float32)
X_test = X_test.astype(np.float32)

y_train = y_train.astype(np.int64)
y_test = y_test.astype(np.int64)
```

### INFORMACIoN

```python
print("Datos cargados correctamente.")
print()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

print()

print("Actividades presentes:")
print(np.unique(y_train))
```

### GUARDAR

```python
np.savez_compressed(
    ARCHIVO_SALIDA,
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test
)

print(f"\nDataset guardado en: {ARCHIVO_SALIDA}")
```
